```python
import time
from pop import Pilot
```

```python
Car=Pilot.AutoCar()

#앞바퀴 (-1 ~ 1)
Car.steering = -0.5
time.sleep(1)
Car.steering = 0.5
```

```python
#전진 후진
Car.forward(10)
time.sleep(3)
Car.backward(10)
time.sleep(3)
Car.stop()
```

```python
#속도조절
Car.setSpeed(50)
Car.forward(50)
time.sleep(1)
Car.backward(50)
time.sleep(1)
Car.stop()
```

```python
#가상조이스틱
Car.joystick()
```

```python
#카메라
Car.camPan(0)
time.sleep(1)
Car.camPan(90)
time.sleep(1)
Car.camTilt(90)
time.sleep(1)
Car.camTilt(0)
```

```python
#각속도
value = Car.getGyro()
print(value)
time.sleep(1)
value=Car.getGyro('x')
print("x : ", value)
value=Car.getGyro('y')
print("y : ", value)
value=Car.getGyro('z')
print("z : ", value)
```

```python
#가속도
value = Car.getAccel()
print(value)
time.sleep(1)
value=Car.getAccel('x')
print("x : ", value)
value=Car.getAccel('y')
print("y : ", value)
value=Car.getAccel('z')
print("z : ", value)
```
