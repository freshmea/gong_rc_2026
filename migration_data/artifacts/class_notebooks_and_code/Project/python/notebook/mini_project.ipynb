```python
from pop import LiDAR, Pilot
import time
```

```python
from IPython.display import clear_output
```

```python
Car = Pilot.AutoCar()
```

```python
lidar = LiDAR.Rplidar()
lidar.connect()
```

```python
lidar.startMotor()
```

```python
vectors = lidar.getVectors()
for v in vectors:
    print(v[0],v[1],v[2])
    clear_output()
```

```python
Car.setSpeed(99)
Car.steering = 0
```

```python
Car.stop()
```

```python
#vector [각도. 거리. 데이터신뢰도]

Car.forward(90)
while True:
    vectors = lidar.getVectors()
    for v in vectors:
        if v[0] >= 300 or v[0] <= 60: #전방 120도
         #if 170 <= v[0] <= 190:
            if v[1] <= 450: # 20cm 이하
                print(v[0], " good ", v[1])

                break
                
            else:
                print(v[0], " bad ", v[1])
                
    
    if v[1] <= 450:
        print("break : ", v[0], " ", v[1])
        break

Car.stop()
    
```
